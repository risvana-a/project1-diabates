# classification and the split

```python
feature_cols = [
    'age',
    'pregnancies',
    'glucose',
    'blood_pressure',
    'skin_thickness',
    'insulin',
    'bmi',
    'diabetes_pedigree',
    'bmi category',
    'glucose category'
]
```

separate the feature from the target

```python
X = df[feature_cols]
Y = df["outcome"]
```

```python
X = pd.get_dummies(X,drop_first=True)
```
