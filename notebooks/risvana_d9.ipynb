# explain how knn reaches a decision

KNN makes a decision by finding the K closest data points and choosing the class that appears most often among them.


Euclidean distance = distance between two points.

# why scaling matters more hear

Scaling is important in KNN because KNN depends on distance, and scaling prevents large-number features from dominating.

score all three with the same function

```python
from sklearn.linear_model import LogisticRegression
def evaluate_model(name, model, X_test, y_test):
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    cm = confusion_matrix(y_test, y_pred)
    tn, fp, fn, tp = cm.ravel()

    print(name)
    print("Accuracy:", round(acc, 4))
    print("Confusion matrix:", cm.tolist())
    print("Diabetics found :", tp, "of", tp + fn)
    print("Missed          :", fn)

    return { "model": name,"accuracy": round(acc, 4),"found": int(tp),"missed": int(fn),"false_alarm": int(fp)}

logreg = LogisticRegression(max_iter=1000).fit(x_train_scaled, y_train)
knn_best = KNeighborsClassifier(n_neighbors=25).fit(x_train_scaled, y_train)

r_log = evaluate_model("LogisticRegression(baseline)", logreg,x_test_scaled,y_test)
r_tree= evaluate_model("Decision Tree (max_dept=3)",tree_best,x_test,y_test)
r_knn = evaluate_model("KNN (k=25)",knn_best,x_test_scaled,y_test)
```
