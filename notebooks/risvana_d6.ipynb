# EVALUATION MATRIX

bulding the confusion matrix

```python
from sklearn.metrics import confusion_matrix 
cm=confusion_matrix(Y_test, Y_pred)
print(cm)
tn, fp, fn, tp = cm.ravel()
print("False positives:", fp)

print("False negatives:", fn)

print("True positives :", tp)
print("True negatives :", tn)
```

# all foure cells, in pattent terms

| Cell   |   Count | What it means for real patients                          |
| ------ | ------: | -------------------------------------------------------- |
| **TN** | **127** | Predicted **no diabetes** and actually **no diabetes**   |
| **FP** |   **8** | Predicted **diabetes**, but actually **no diabetes**     |
| **FN** |  **39** | Predicted **no diabetes**, but actually **had diabetes** |
| **TP** |  **16** | Predicted **diabetes** and actually **had diabetes**     |

```python
missed_rate = fn / (fn + tp)

print("Diabetic patients in the test set :", fn+tp) 
print("Found by the model                :", tp)
print("Missed by the model               :",fn)
print("Proportion missed                 :",round(missed_rate,4))
```

writte a reusable  evaluation function

```python
def evaluate_model(name, model, X_test, y_test):
    Y_pred = model.predict(X_test)
    acc =accuracy_score(y_test, Y_pred)
    cm = confusion_matrix(y_test, Y_pred)
    tn, fp, fn, tp = cm.ravel()

    print(name)
    print ("accuracy:", round(acc, 4))
    print("confusion_matrix:", cm.tolist())
    print("dibetics found :", tp, "of" , tp+fp)
    print ("missed:", fn)

    return{"model":name,"accuracy":round(acc,4),"found":tp,"missed":fn}     
```

```python
baseline_result = evaluate_model(
    "logistic regression (baseline)", model, X_test_scaled, Y_test)
```

```python
stability=[]
for seed in [0,7,42,99,2026]:
    xtr, xte, ytr, yte = train_test_split(
        X,Y,test_size=0.2, random_state=seed, stratify=Y)
    sc = StandardScaler().fit(xtr)
    m = LogisticRegression(max_iter=1000).fit(sc.transform(xtr),ytr)
    pred = m.predict(sc.transform(xte))
    cmx= confusion_matrix(yte,pred)
    stability.append({
        "random_state":seed,
        "accuracy":round(accuracy_score(yte,pred),4),
        "missed":int(cmx[1,0]),
    })
stab = pd.DataFrame(stability)
print(stab.to_string(index=False))   
```
