# comparing all tree


```python
from sklearn.linear_model import LogisticRegression

def evaluate_model(name, model, X_test, y_test):
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)

    cm = confusion_matrix(y_test, y_pred)
    tn, fp, fn, tp = cm.ravel()

    print(name)
    print("Accuracy       :", round(acc, 4))
    print("Confusion matrix:", cm.tolist())
    print("Diabetics found :", tp, "of", tp + fn)
    print("Missed          :", fn)

    return {
        "model": name,
        "accuracy": round(acc, 4),
        "found": int(tp),
        "missed": int(fn),
        "false_alarm": int(fp)
    }


logreg = LogisticRegression(max_iter=1000).fit(
    x_train_scaled, y_train
)

knn_best = KNeighborsClassifier(n_neighbors=25).fit(
    x_train_scaled, y_train
)
```

```python
r_log=evaluate_model('LogisticRegression(baseline)',logreg,x_test_scaled,y_test)
r_tree=evaluate_model('DecisionTree(max_depth=3)',tree_best,x_test,y_test)
r_knn=evaluate_model('knn(k=25)',knn_best,x_test_scaled,y_test)
```

```python
result=pd.DataFrame([r_log,r_tree,r_knn])
result.sort_values('accuracy',ascending=False)
```

```python
result=pd.DataFrame([r_log,r_tree,r_knn])
result.sort_values('found',ascending=False)
```

```python
def cm_row(name,model,xt):
   cm=confusion_matrix(y_test,model.predict(xt))
   tn,fp,fn,tp=cm.ravel()
   return{"model":name,"correct_negatives":int(tn),"false_alarams":int(fp),"missed_patients":int(fn),"found_patients":int(tp),"accuracy": round(model.score(xt, y_test), 4)} 
pd.DataFrame([
    cm_row("Logistic Regression",logreg, x_test_scaled),
    cm_row("Decision Tree (d=3)", tree_best, x_test),
    cm_row("KNN (k=25)", knn_best, x_test_scaled),
    cm_row("Decision Tree (unlimited)", tree_full, x_test)
])
```
